```python
import pandas as pd

file_path = 'C:/Users/user/Desktop/데이터들/Data/real_estate_final.csv'

real_estate = pd.read_csv(
    file_path,
    encoding='cp949'
)
real_estate.head()
```

```python
real_estate = real_estate.replace(r'^\s*$', pd.NA, regex=True)
```

```python
print(real_estate.isna().sum())
```

```python
numeric_cols = [
    'year',
    'transaction_amount',
    'deposit',
    'rent'
]

for col in numeric_cols:
    real_estate[col] = pd.to_numeric(
        real_estate[col],
        errors='coerce'
    )
```

```python
print(real_estate.dtypes)
```

```python
output_path = 'C:/Users/user/Desktop/데이터들/Data/real_estate_final_clean.csv'

real_estate.to_csv(
    output_path,
    index=False,
    encoding='utf-8-sig'
)
```

```python
import pandas as pd
from sqlalchemy import create_engine
from tqdm import tqdm

# MySQL 연결
engine = create_engine(
    "mysql+pymysql://root:class1234@localhost/subway_realestate?charset=utf8mb4"
)

# CSV 읽기
print("CSV 읽는 중...")

df = pd.read_csv(
    'real_estate_final.csv',
    encoding="EUC-KR"
)

print("CSV 읽기 완료")
print("총 행 수:", len(df))

# INSERT
chunk_size = 5000
total = len(df)

print("MySQL INSERT 시작...")

for start in tqdm(
    range(0, total, chunk_size),
    desc="INSERT 진행률",
    unit="chunk"
):

    end = min(start + chunk_size, total)

    df.iloc[start:end].to_sql(
        name="real_estate_temp",
        con=engine,
        if_exists="append",
        index=False,
        method="multi"
    )

print("INSERT 완료")
print("총 행 수:", len(df))
```
