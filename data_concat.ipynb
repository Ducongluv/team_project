```python
import pandas as pd


# 매매
# 아파트, 오피스텔, 단독다가구, 연립 다세대 파일 열기
apt = pd.read_excel('C:/Users/user/Desktop/데이터들/Data/아파트_매매_33역.xlsx')
op = pd.read_excel('C:/Users/user/Desktop/데이터들/Data/오피스텔_매매_33역.xlsx')
dan = pd.read_excel('C:/Users/user/Desktop/데이터들/Data/단독다가구_매매_33역.xlsx')
yun = pd.read_excel('C:/Users/user/Desktop/데이터들/Data/연립다세대_매매_33역.xlsx')

# 전,월세
jeon = pd.read_excel('C:/Users/user/Desktop/데이터들/Data/전세_2호선_33역_통일.xlsx')
wol = pd.read_excel('C:/Users/user/Desktop/데이터들/Data/월세_2호선_33역_통일.xlsx')
```

```python
# 데이터 통합 준비

# 거래유형 추가
apt['거래유형'] = '매매'
op['거래유형'] = '매매'
dan['거래유형'] = '매매'
yun['거래유형'] = '매매'

for df in [apt, op, dan, yun]:
    df['보증금'] = pd.NA
    df['임대료'] = pd.NA

apt.head(5)
```

```python
dan.head()
```

```python
# 전, 월세 거래유형 생성
jeon['거래유형'] = '전세'
wol['거래유형'] = '월세'

jeon['거래금액'] = pd.NA
wol['거래금액'] = pd.NA
```

```python
jeon.rename(columns={'전세보증금': '보증금'}, inplace= True)

jeon.head(10)
```

```python
wol.head(5)
```

```python
wol['주택유형'].unique()
```

```python
real_estate = pd.concat(
    [
        apt,
        op,
        dan,
        yun,
        jeon,
        wol
    ],
    ignore_index=True
)
```

```python
print(real_estate.head())
print(real_estate.shape)

print(real_estate['주택유형'].unique())
print(real_estate['거래유형'].unique())
```

```python
# 주택 유형이 아파트 오피스텔 단독다가구 연립다세대로만 나와야 하는데 단독, 연립, 다가구, 다세대가 각각 따로 있음 -> 하나로 합치기
dan['주택유형'] = (
    dan['주택유형']
    .astype(str)
    .str.strip()
    .replace({
        '단독': '단독다가구',
        '다가구': '단독다가구'
    })
)

yun['주택유형'] = (
    yun['주택유형']
    .astype(str)
    .str.strip()
    .replace({
        '연립': '연립다세대',
        '다세대': '연립다세대'
    })
)
```

```python
print('단독다가구:', dan['주택유형'].unique())
print('연립다세대:', yun['주택유형'].unique())
```

```python
real_estate1 = pd.concat(
    [
        apt,
        op,
        dan,
        yun,
        jeon,
        wol
    ],
    ignore_index=True
)
```

```python
print(real_estate1.head())
print(real_estate1.shape)

print(real_estate1['주택유형'].unique())
print(real_estate1['거래유형'].unique())
```

```python
real_estate1.to_csv(
    'C:/Users/user/Desktop/데이터들/Data/부동산통합.csv',
    index=False,
    encoding='utf-8-sig'
)
```
